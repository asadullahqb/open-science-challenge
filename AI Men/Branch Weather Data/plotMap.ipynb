```python
import pandas as pd
import folium
from folium import plugins

# Load the datasets
submission_df = pd.read_csv('submission.csv')
training_df = pd.read_csv('Training_data_uhi_index.csv')

# Create a base map centered on the mean coordinates
center_lat = (submission_df['Latitude'].mean() + training_df['Latitude'].mean()) / 2
center_lon = (submission_df['Longitude'].mean() + training_df['Longitude'].mean()) / 2
m = folium.Map(location=[center_lat, center_lon], zoom_start=4)

# Add markers for submission data points
for idx, row in submission_df.iterrows():
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=3,
        color='red',
        fill=True,
        popup=f"Submission Point {idx}",
    ).add_to(m)

# Add markers for training data points
for idx, row in training_df.iterrows():
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=3,
        color='blue',
        fill=True,
        popup=f"Training Point {idx}",
    ).add_to(m)

# Add a legend
legend_html = '''
<div style="position: fixed; 
            bottom: 50px; right: 50px; width: 150px; height: 90px; 
            border:2px solid grey; z-index:9999; background-color:white;
            opacity:0.8;
            font-size:12px;
            padding: 10px">
            <p><i class="fa fa-circle" style="color:red"></i> Submission Points</p>
            <p><i class="fa fa-circle" style="color:blue"></i> Training Points</p>
</div>
'''
m.get_root().html.add_child(folium.Element(legend_html))

# Save the map to an HTML file
m.save('weather_points_map.html')

# Display the map in the notebook
display(m)
```
